1. Get all orders over 1000, status 'shipped', country either 'India' or 'US'.

```python
df = df.filter((col('amount') > 1000) & (col('status') == 'shipped') & (col('country').isin('India', 'US')))
```

2. Filter employees who are either in Sales with salary above 50000, OR Engineering with salary above 70000.

```python
df = df.filter(
    ((col('department') == 'Sales') & (col('salary') > 50000)) |
    ((col('department') == 'Engineering') & (col('salary') > 70000))
)
```

3. For each department, find the employee with the highest salary (return just the top row per department).

```python
df = df.withColumn('rnk', row_number().over(Window.partitionBy('user_id').orderBy(col('login_timestamp').desc()))).filter(col('rnk') == 3)
```

4. Find the 3rd most recent login per user from a logins table.

```python
df = df.withColumn('rnk', row_number().over(Window.partitionBy('user_id').orderBy(col('login_timestamp').desc()))).filter(col('rnk') == 3)
```

5. Find customers whose most recent order amount was lower than their average order amount.

```python
avg_df = df.groupBy('cust_id').agg(avg('amount').alias('avg_amt'))

recent_df = df.withColumn('rnk', row_number().over(Window.partitionBy('cust_id').orderBy(col('order_date').desc()))).filter(col('rnk') == 1)

result = recent_df.join(avg_df, on='cust_id').filter(col('amount') < col('avg_amt'))
```
